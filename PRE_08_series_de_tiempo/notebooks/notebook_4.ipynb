# 4. Modelo multiplicativo: regresion sobre log(y)

Si la estacionalidad es proporcional al nivel, en logaritmos se vuelve aditiva.

```python
%run functions.ipynb
```

```python
from sklearn.linear_model import LinearRegression

df = load_data()
df = add_linear_trend_component(df)
df = add_month_component(df)

dummies = pd.get_dummies(df.month, prefix="month", drop_first=True, dtype=int)
df = pd.concat([df, dummies], axis=1)
df["log_yt"] = np.log(df.yt_true)

x_columns = ["trend"] + list(dummies.columns)
X_complete, y_complete, X_train, y_train, X_test, y_test = train_test_split(df, x_columns, "log_yt")

model = LinearRegression().fit(X_train, y_train)
df["yt_pred_log_trend_month"] = np.exp(model.predict(X_complete))

plot_time_series(df[["yt_true", "yt_pred_log_trend_month"]])
```

```python
metrics = compute_evaluation_metrics(df)
save_forecasts(df)
save_metrics(metrics)
metrics
```
